```python
import mne
from mne.time_frequency import tfr_morlet
```

```python
epochs_for_tfr = mne.read_epochs("S1_epo.fif")
psd = epochs_for_tfr.compute_psd(fmin=2, fmax=20)
```

```python
psd.plot()
```

```python
freqs = list(range(1,20))
tfr = epochs_for_tfr[1200].compute_tfr(method="morlet", freqs=freqs, n_cycles=2, average=True)
```

```python
tfr.plot()
```
